```python
import requests
import json

# Define o endpoint do OpenAlex solicitando uma amostra de 50 trabalhos (works)
url = "https://api.openalex.org/works?per-page=50"

# Executa a requisição GET na API
response = requests.get(url)

if response.status_code == 200:
    # O OpenAlex envia metadados na raiz. Os dados reais estão na lista 'results'
    resultados = response.json().get("results", [])
    
    # Converte os resultados para o formato JSON Lines (um registro por linha)
    json_lines = "\n".join([json.dumps(registro, ensure_ascii=False) for registro in resultados])
    
    # Define o caminho relativo para a pasta que você criou
    caminho_arquivo = "Files/01_Bronze/amostra_works.json"
    
    # Usa o utilitário nativo do Fabric para gravar o arquivo físico no Lakehouse
    mssparkutils.fs.put(caminho_arquivo, json_lines, True)
    
    print(f"Sucesso! {len(resultados)} registros foram extraídos e salvos em: {caminho_arquivo}")
else:
    print(f"Falha na requisição da API. Código de erro HTTP: {response.status_code}")
```
